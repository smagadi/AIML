```python
from phi.agent import Agent
from phi.model.openai import OpenAIChat
from phi.tools.googlesearch import GoogleSearch
from phi.tools.yfinance import YFinanceTools
import os

```

```python
from google.colab import userdata
import openai

key =userdata.get('OPENAI')

```

```python
# Retrieve the API key from Colab Secrets
# Try setting the API key as an environment variable:
os.environ["OPENAI_API_KEY"] = userdata.get('OPENAI')
```

```python
## create senti Agent
sentiment_agent = Agent(
    name="Sentiment Agent",
    role="Search and Interpret News Articles",
    model=OpenAIChat(id="gpt-4o-mini"),
    tools=[GoogleSearch()],
    instructions=[
        "Find relevant news articles for each company and analyze the sentimenets",
        "Provide sentiment score between 1 to 10 (1 being lowest and 10 being highest) with reasoning and sources"
        "Cite your resources and links"
    ],
    show_tool_calls=True,
    markdown=True,
)
```

```python
# prompt: #create a finance agent that gets finance data and interpret trends

tech_analysis_finance_agent = Agent(
    name="Tech Analysis Finance Agent",
    role="Analyze the Techical parameters of the stock using candlestick and recommend a short term pricing ,state the reasons",
    model=OpenAIChat(id="gpt-4o-mini"),  # You can use a different model if needed
    tools=[YFinanceTools(stock_price =True ,analyst_recommendations= True ,company_info=True)],
    instructions=[
            "Analyse the key Techical Analysis parameters and present in a table format "
            "Focus on trends and present the data in tables ",
            "Understand the key signals and present the recommendations"
            "Provide the candlestick analysis"

    ],
    show_tool_calls=True,
    markdown=True,
)

```
