```python
from phi.agent import Agent
from phi.model.openai import OpenAIChat
from phi.tools.googlesearch import GoogleSearch
from phi.tools.yfinance import YFinanceTools
import os

```

```python
from google.colab import userdata
import openai

key =userdata.get('OPENAI')

```

```python
# Retrieve the API key from Colab Secrets
# Try setting the API key as an environment variable:
os.environ["OPENAI_API_KEY"] = userdata.get('OPENAI')
```

```python
## create senti Agent
sentiment_agent = Agent(
    name="Sentiment Agent",
    role="Search and Interpret News Articles",
    model=OpenAIChat(id="gpt-4o-mini"),
    tools=[GoogleSearch()],
    instructions=[
        "Find relevant news articles for each company and analyze the sentimenets",
        "Provide sentiment score between 1 to 10 (1 being lowest and 10 being highest) with reasoning and sources"
        "Cite your resources and links"
    ],
    show_tool_calls=True,
    markdown=True,
)
```

```python
# prompt: #create a finance agent that gets finance data and interpret trends

finance_agent = Agent(
    name="Finance Agent",
    role="Analyze Financial Data and Trends",
    model=OpenAIChat(id="gpt-4o-mini"),  # You can use a different model if needed
    tools=[YFinanceTools(stock_price =True ,analyst_recommendations= True ,company_info=True)],
    instructions=[
        "Retrieve Stock prices,analyst recommendations and key financial data ",
        "Analyse the key Techical Analysis parameters and present in a table"
        "Focus on trends and present the data in tables with key insights.",

    ],
    show_tool_calls=True,
    markdown=True,
)

```

```python
# prompt: #create a analysis agent that gets finance data and interpret trends

analysis_agent = Agent(
    name="Analysis Agent",
    role="Ensure thorougness and draw conclusions",
    model=OpenAIChat(id="gpt-4o-mini"),  # You can use a different model if needed
    instructions=[
        "Check outputs for accuracy and completness",
        "Use data from sentiment_agent and finance_agent"
        " Present the Short term ,medium term and long term view based on Technical Analysis"
        "Create a structured table format and present the data "
    ],
    show_tool_calls=True,
    markdown=True,
)

```

```python
# prompt : #create an agent team combines the expertise of all agents to provide a cohesive
#       :well supported response ,incluse data sources and citations with reasoning
#.       :presnet all data in strutured table format ,explain methodology arrived at sentiment score

agent_team = Agent(
    model=OpenAIChat(id="gpt-4o-mini"),
    name="Agent Team",
    instructions=[
        "combines the expertise of all agents to provide a cohesive"
        " Always include well supported response , data sources and citations with"
        " Present all data in strutured table format ,explain methodology arrived at sentiment score"

    ],
    show_tool_calls=True,
    markdown=True,
)

```

```python
# Final promot
agent_team.print_response(
    "Help me to understand about the INFY stock symbol of NSE , what is the long and short term view , what is the sentiments ?  "
    ,stream=True)

```
