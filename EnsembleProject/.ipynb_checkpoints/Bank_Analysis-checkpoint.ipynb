## Deliverable – 2 (Prepare the data for analytics)

```python
## Data Cleaning 
```

```python
def get_dummy_from_bool(row, column_name):
    ''' Returns 0 if value in column_name is no, returns 1 if value in column_name is yes'''
    return 1 if row[column_name] == 'yes' else 0

def get_correct_values(row, column_name, threshold, df):
    ''' Returns mean value if value in column_name is above threshold'''
    if row[column_name] <= threshold:
        return row[column_name]
    else:
        mean = df[df[column_name] <= threshold][column_name].mean()
        return mean
```

```python
def clean_data(df):
    
    
    cleaned_df = ds.copy()
    
    #convert columns containing 'yes' and 'no' values to boolean variables and drop original columns
    bool_columns = ['default', 'housing', 'loan', 'Target']
    for bool_col in bool_columns:
        cleaned_df[bool_col + '_bool'] = df.apply(lambda row: get_dummy_from_bool(row, bool_col),axis=1)
    
    cleaned_df = cleaned_df.drop(columns = bool_columns)
    
    #convert categorical columns to dummies
    cat_columns = ['job', 'marital', 'education', 'contact', 'month', 'poutcome']
    
    for col in  cat_columns:
        cleaned_df = pd.concat([cleaned_df.drop(col, axis=1),
                                pd.get_dummies(cleaned_df[col], prefix=col, prefix_sep='_',
                                               drop_first=True, dummy_na=False)], axis=1)
    
    #drop irrelevant columns
    cleaned_df = cleaned_df.drop(columns = ['pdays'])
    
    #impute incorrect values and drop original columns
    cleaned_df['campaign_cleaned'] = df.apply(lambda row: get_correct_values(row, 'campaign', 34, cleaned_df),axis=1)
    cleaned_df['previous_cleaned'] = df.apply(lambda row: get_correct_values(row, 'previous', 34, cleaned_df),axis=1)
    
    cleaned_df = cleaned_df.drop(columns = ['campaign', 'previous'])
    
    return cleaned_df
```

```python
cleaned_df = clean_data(ds)
cleaned_df.head()
```

## Deliverable – 2 Create the training set and test set in ration of 70:30

```python
X = cleaned_df.drop("Target_bool" , axis=1)
y = cleaned_df.pop("Target_bool")
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=.30, random_state=1)
```

## Deliverable – 3 (create the Base model)

```python
# Build Decision Tree
```

```python
dTree = DecisionTreeClassifier(criterion = 'entropy', random_state=1)
dTree.fit(X_train, y_train)
y_pred = dTree.predict(X_test)
```

```python
## Scoring 
```

```python


print("Accuracy of base DT:",metrics.accuracy_score(y_test,y_pred))
print("Confusion Metrix of of base DT:\n",metrics.confusion_matrix(y_test,y_pred))
print("Recall of base DT:\n",metrics.recall_score(y_test,y_pred))

```

## Deliverable 3 - Base Model Performance

### As the Target value is skewed we will use the Recall to evalauate model

### Advantages of Decision Tree
### 1. Decision trees implicitly perform variable screening or feature selection.
### 2. Can handle both numerical and categorical data. Can also handle multi-output problems.
### 3. Decision trees require relatively little effort from users for data preparation.
### 4. Nonlinear relationships between parameters do not affect tree performance.

### Disadvantages of Decison Tree
### 1. Decision-tree learners can create over-complex trees that do not generalize the data well. This is called overfitting.
### 2. Decision trees can be unstable because small variations in the data might result in a completely different tree being generated. This is called variance, which needs to be lowered by methods like bagging and boosting.
